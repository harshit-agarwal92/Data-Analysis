```python
import json
t = '''[
    {
        "shop": "FreshMart",
        "open": true,
        "owner": null,
        "items": [
            {"name": "Rice", "price": 60, "qty": 10},
            {"name": "Oil", "price": 150, "qty": 4},
            {"name": "Tea", "price": 120, "qty": 6}
        ]
    }
]'''
# 1. Load JSON into Python object and print its type
data = json.loads(t)
print(type(data))
```

```python
# 2. Print the name of the second item and number of items
print(data[0]["items"][1]["name"])
print(len(data[0]["items"]))
```

```python
# 3. Print the value of owner
print(data[0]["owner"])
```

```python
# 4. Convert the object back to JSON with indentation 2
json_text = json.dumps(data, indent=2)
print(json_text)
```
