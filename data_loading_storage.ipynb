```python
import sqlite3


conn = sqlite3.connect("stduents.db")
cursor = conn.cursor()


cursor.execute("DROP TABLE IF EXISTS student_table")


cursor.execute("""
CREATE TABLE student_table(
    NAME TEXT,
    ROLL_NO INTEGER,
    MARKS INTEGER
)
""")


cursor.execute("INSERT INTO student_table VALUES(?, ?, ?)", ("Aastik", 1, 99))
cursor.execute("INSERT INTO student_table VALUES('Disha', 14, 99)")
cursor.execute("INSERT INTO student_table VALUES('Harshit', 18, 100)")

conn.commit()


cursor.execute("SELECT * FROM student_table")
for row in cursor.fetchall():
    print(row)

conn.close()
```
