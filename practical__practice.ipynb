```python
#create a csv with no header and 5 columns.Load it with your own column names and make the first column the index.
import pandas as pd
data = [
    [101, "Aman", 20, "Mathura", 85],
    [102, "Rahul", 21, "Agra", None],
    [103, "Priya", 19, "Delhi", 88],
    [104, "Neha", 20, "Noida", 92],
    [105, "Rohit", 22, "Jaipur", 80]
]
df = pd.DataFrame(data)
df.to_csv("data.csv", index=False, header=False)
# Load CSV with custom column names and first column as index
df = pd.read_csv(
    "data.csv",
    header=None,
    names=["ID", "Name", "Age", "City", "Marks"],
    index_col=0
)

print(df)
```

```python
#wirte a file using ; as the delimiter and -999 fro missing values. Read it so -999 becomes NaN.
rf=pd.DataFrame(data)
rf.to_csv( "data.csv", sep=";", index=False, header=False, na_rep="-999" )
rf = pd.read_csv( "data.csv", sep=";", header=None, names=["ID", "Name", "Age", "City", "Marks"], na_values=["-999"] )
print(rf)
```

```python
#Read a big csv in chunks of 2000 rows and compute the mean of one column per key acorss all chunks.
chunks=pd.read_csv("/content/Titanic-Dataset.csv",chunksize=2000)
totalsum={}
totalcount={}
for chunk in chunks:
  grouped=chunk.groupby("Sex")["Age"].agg(["sum","count"])

  for key, row in grouped.iterrows():
    totalsum[key]=totalsum.get(key,0)+row["sum"]
    totalcount[key]=totalcount.get(key,0)+row["count"]
for key in totalsum:
  mean=totalsum[key]/totalcount[key]
  print(key,mean)
```
